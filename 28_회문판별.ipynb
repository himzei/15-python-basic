```python
# 회문 판별하기 
# 반복문으로 문자 검사하기 

word = input("단어를 입력하세요 : ")

is_palindrome = True 

for i in range(len(word) // 2):
    if word[i] != word[-1 -i]: 
        is_palindrome = False
        break

print(is_palindrome)
```

```python
# 시퀀스 뒤집기로 문자 검사하기 
word = input("단어를 입력하세요 : ")

print(word == word[ : :-1])
```

```python
# 리스트와 reversed 사용하기 

word = "level"

list(word) == list(reversed(word))
```

```python
word = "level"

word == "".join(reversed(word))
```

```python
# N-gram 만들기
# 반복문으로 N-gram 출력하기 

test = "hello"

for i in range(len(test) - 1):
    print(test[i], test[i+1], sep="")
```

```python
text = "this is python script"
words = text.split()

for i in range(len(words) - 1):
    print(words[i], words[i+1])
```

```python
# zip으로 2-gram 만들기
text = "hello"

two_gram = zip(text, text[1:])
# list(two_gram)

for i in two_gram: 
    print(i[0], i[1], sep="")
```
