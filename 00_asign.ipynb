```python
# 문제 1.
# input() 함수를 이용하여 국어, 영어, 수학, 과학, 컴퓨터 점수를 한 줄로 입력받으세요.
# 입력된 점수는 콤마 ,로 구분되어 있습니다.
# 입력받은 점수들의 평균 점수를 출력하는 프로그램을 작성하세요

# 입력값 : 
# 80,90,70,85,95

# 출력값 : 
# 평균 점수: 84.0

# 요구사항 
# 1. input() 함수를 사용하세요.
# 2. 점수는 콤마 ,로 구분하여 입력받으세요.
# 3. split()을 사용하여 점수를 분리하세요.

# -----------

국어, 영어, 수학, 과학, 컴퓨터 = map(int,input("점수를 입력하세요:").split(","))
print(f"평균 점수: {(국어+영어+수학+과학+컴퓨터)/5}")
```

```python
# 문제 2.
# input() 함수를 이용하여 년, 월, 일, 시, 분, 초가 입력을 하고 
# 제시된 결과값을 출력하는 프로그램을 작상하세요 

# 입력값 : 
# 2026 5 13 16 40 1

# 출력값 : 
# 2026/05/13T16:40:01

# 요구사항
# 1. print() 함수의 매개변수를 활용하세요 


# ----------
년,월,일,시,분,초 = map(int,input("년, 월, 일, 시, 분, 초가 입력 :").split())
# print(f"{년}/{월:02}/{일:02}T{시:02}:{분:02}:{초:02}")
print(년, f"{월:02}", 일, sep="/", end="")
print(f"T{시}", 분, f"{초:02}", sep=":")
```

```python
# 문제 3. 
# input() 함수를 이용하여 국어, 영어, 수학, 과학, 컴퓨터 점수를 입력받아 
# 한 과목이라도 60점 미만이면 False 가 나오도록 프로그램을 작성하세요 

# 입력값 : 
# 70 80 55 60 90

# 출력값 : 
# False 

# 요구사항
# 1. 비교연산자와 논리연산자를 사용하세요 


# ---------
국어, 영어, 수학, 과학, 컴퓨터 = map(int, input().split())

print(국어 >= 60 and 영어 >= 60 and 수학 >= 60 and 과학 >= 60 and 컴퓨터 >= 60)
```

```python
# 문제 4. 
# 튜플 언패킹을 이용하여 각각의 값을 변수에 저장한 뒤 출력하시오

# member = ("hong", "홍길동", 25, "대구")

# 출력값 : 
# 아이디: hong
# 이름: 홍길동
# 나이: 25
# 지역: 대구


# ----------
member = ("hong", "홍길동", 25, "대구")

아이디, 이름, 나이, 지역 = member

print(f"아이디: {아이디}")
print(f"이름: {이름}")
print(f"나이: {나이}")
print(f"지역: {지역}")
```

```python
# 문제 5. 
# range()로 숫자 리스트 만들기
# 다음 조건을 만족하는 리스트를 각각 만드세요.

# 1. 1부터 10까지의 숫자 리스트
# 2. 0부터 20까지의 짝수 리스트
# 3. 1부터 19까지의 홀수 리스트
# 4. 10부터 1까지 거꾸로 된 리스트
# 5. 5부터 50까지 5씩 증가하는 리스트



# ----------
list1 = list(range(1, 11))
print(list1)

list2 = list(range(0, 21, 2))
print(list2)

list3 = list(range(1, 20, 2))
print(list3)

list4 = list(range(10, 0, -1))
print(list4)

list5 = list(range(5, 51, 5))
print(list5)
```

```python
# 문제 6. 
# input 함수로 입력을 받아서 
# 튜플로 출력하는 프로그램을 만드세요 

# 입력 :
# 시작값: 3
# 끝값: 15
# 증가값: 3


# 출력값 : 
# (3, 6, 9, 12)

# 요구사항

# 1. input(), range(), tuple(), int() 사용하세요
# ----------
시작값 = int(input("시작값: "))
끝값 = int(input("끝값: "))
증가값 = int(input("증가값: "))

result = tuple(range(시작값, 끝값, 증가값))

print(result)
```

```python
# 문제 7. 
# 이메일 주소를 분리하여 출력하는 프로그램을 만드세요 

# 입력값 : 
# himzei@gmail.com

# 출력값 : 
# 아이디는 himzei 입니다. 
# 도메인은 gmail.com 입니다. 


아이디, 도메인 = input().split("@")

print(f"아이디는 {아이디} 입니다.\n도메인은 {도메인} 입니다.")
```
