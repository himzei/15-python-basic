```python
students = {
    "김철수": [90, 80, 70],
    "이영희": [100, 95, 90],
    "박민수": [60, 70, 65]
}

print("===== 학생 성적 결과 =====")

print(students.items())/

total = 0

for name, scores in students.items():
    avg = sum(scores) / len(scores)

    print()
    print("이름:", name)
    print("점수:", scores)
    print("평균:", avg)

    total += avg

overall_avg = total / len(students)

print()
print("===== 전체 결과 =====")
print("전체 학생 수:", len(students), "명")
print("전체 평균:", overall_avg)

```

```python
# 문제 5.

# for문을 이용하여 학생 정보 5명을 입력받고 출력하는 프로그램을 작성하세요.

# 입력항목 
# 이름
# 나이
# 학년

# 요구사항

# 1. for문을 사용하세요.
# 2. 학생 정보는 딕셔너리로 저장하세요.
# 3. 학생 5명의 정보는 리스트에 저장하세요.
# 4. 입력이 끝나면 전체 학생 정보를 출력하세요.

students = []

for i in range(5):
    print(i+1, "번째 학생 정보 입력")

    name = input("이름: ")
    age = input("나이: ")
    grade = input("학년: ")

    student = {
        "이름": name,
        "나이": age,
        "학년": grade
    }

    students.append(student)
    print()

print("===== 전체 학생 정보 =====")

for s in students:
    print(s)
```

```python
# 문제 1. 
# 두 수 사이의 숫자 중  3과 7과 끝나지 않는 숫자 출력하기 

# 입력값 
# 3, 50

# 첫 번째 정수와 두 번째 정수 사이의 숫자 중 3 또는 7로 끝나지 않늣 숫자가 출력되게 만드세요 

# # 1)
f, s = map(int, input().split(','))

for i in range(f, s+1):
    if i % 10 == 3 or i % 10 == 7:
        continue
    print(i)

# ---
```

```python
# 문제 2. 
# 표준 입력으로 역삼각형의 높이가 입력됩니다. 입력된 높이만큼 역삼각형의 "*"을 출력하는 프로그램을 작성하세요 

# 입력갑
# 5

# 출력값 
# *********
#  *******
#   *****
#    ***
#     *

# 요구사항:
# 중첩 for를 사용

# 2)
n = int(input())

for i in range(n):
    for j in range(i):
        print(' ', end='')
    for j in range(2*(n-i)-1):
        print('*', end='')
    print()

# ---
```

```python
# 문제 3. 
# 입력값을 참조하여 첫 번째 정수 부터 두 번째 정수까지 숫자를 역으로 출력하면서 
# 3의 배수일때는 Fizz, 7의 배수일때는 Buzz, 그외는 숫자를 출력하세요 

# 입력값 
# 100 10

# 3)
f, s = map(int, input().split())
if f<s:
    print('잘못된 범위')
else:
    for i in range(f, s-1 , -1):
        if i % 3 == 0 and i % 7 ==0:
            print('FizzBuzz') 
        elif i % 3 == 0:
            print('Fizz')
        elif i % 7 == 0:
            print('Buzz')
        else:
            print(i)

# ---
```
