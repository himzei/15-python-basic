```python
문제. 1

# 가장 높은 점수, 낮은 점수, 총점, 평균, 분산을 구하는 프로그램을 만들어주세요 

# 입력예시 
# 70 80 85 38 43 30
# 또는 
# 100 18 38 94 93 12 39

# 공백으로 구분되어 몇개가 될지 모름

# 요구사항 
# 1. 함수이름은 스스로 지정해 주세요 
# 2. 가변 인수를 사용해 주세요 
----------------------------------------
def get_score_calc(*score):
    max_n = max(score)
    min_n = min(score)
    sum_n = sum(score)
    avg_n = sum_n/len(score)
    
    basic_num = 0
    
    for i in score:
        dev_n = (i-avg_n)**2
        basic_num += dev_n
    var_n = basic_num/len(score)

    return max_n, min_n, sum_n, avg_n ,var_n

score = tuple(map(int,input().split()))

max_n, min_n, sum_n, avg_n, e_num = get_score_calc(*score)

print(f'낮은 점수:{min_n:.2f}, 높은 점수:{max_n:.2f}, 총점:{sum_n:.2f}, 평균:{avg_n:.2f}, 분산:{e_num:.2f}')
```

```python
# prices = {
#     "아메리카노": 2500,
#     "카페라떼": 3500,
#     "바닐라라떼": 4000,
#     "초코케이크": 5500,
#     "샌드위치": 6000
# }

# 함수를 호출하면
# order(아메리카노=2, 카페라떼=1)
# order(바닐라라떼=2, 초코케이크=1, 샌드위치=1)
# order(아메리카노=1, 샌드위치=2)

# 실행결과
# 아메리카노 2개: 5000원
# 카페라떼 1개: 3500원
# 총 결제 금액: 8500원
# --------------------
# 바닐라라떼 2개: 8000원
# 초코케이크 1개: 5500원
# 샌드위치 1개: 6000원
# 총 결제 금액: 19500원
# --------------------
# 아메리카노 1개: 2500원
# 등록되지 않은 상품입니다: 허니브레드
# 총 결제 금액: 2500원
# --------------------

# 요구사항 
# 1. 함수 이름은 order로 만드세요.
# 2. 함수는 **kwargs를 사용해야 합니다.
# 3. kwargs의 key는 상품명, value는 주문 수량입니다.

# ===
prices = {
    "아메리카노": 2500,
    "카페라떼": 3500,
    "바닐라라떼": 4000,
    "초코케이크": 5500,
    "샌드위치": 6000
}

def order(**kwargs):
    total_price = 0
    for i in kwargs.keys():
        if i in prices.keys():
            product_price = kwargs[i] * prices[i]
            print(f"{i} {kwargs[i]}개: {product_price}원")
            total_price += product_price
        else:
            print(f"등록되지 않은 상품입니다: {i}")
    print(f"총 결제 금액: {total_price}원")
    print("-" * 20)
    
order(아메리카노=2, 카페라떼=1)
order(바닐라라떼=2, 초코케이크=1, 샌드위치=1)
order(아메리카노=1, 샌드위치=2)
order(아메리카노=1, 허니브레드=2)
```

```python
import random

names = [
    "권석현", "김규원", "김성준", "김태훈", "김형민", "문홍식", "박규민", "박성민", "박진영", "박찬현",
    "박희재", "손성욱", "신재민", "오은진", "원영배","이도영", "이수현", "이원석", "이진욱", "조현우","홍득훈"
]

student_data = {}

for name in names:
    student_data[name] = [random.randint(60, 100) for _ in range(10)]
print(student_data)
```

```python
class Student:
    def __init__(self, name, scores):
        self.name = name
        self.scores = scores

    def total(self):
        return sum(self.scores)
    def average(self):
        return self.total() / len(self.scores)
    def max_score(self):
        return max(self.scores)
    def min_score(self):
        return min(self.scores)
    def is_pass(self):
        if self.average() >= 70:
            return('합격')
        else:
            return('불합격')
    def print_info(self):
        print('이름: ', self.name)
        print('점수: ', self.scores)
        print('총점: ', self.total())
        print(f'평균: {self.average():.2f}')
        print('최고점수: ', self.max_score())
        print('최저점수: ', self.min_score())
        print('결과: ', self.is_pass())
        print('-'* 20)

```

```python
students = {}
for name, scores in student_data.items(): 
    students[name] = Student(name, scores)


```
