```python
# 딕셔너리 표현식 사용하기(딕셔너리 컴프리헨션)

## 1. {키 : 값 for 키, 값 in 딕셔너리}
## 2. dict({ 키 : 값 for 키, 값 in 딕셔너리 })

keys = ["a", "b", "c", "d"]
x = {key:value for key, value in dict.fromkeys(keys).items()}
print(x)
```

```python
keys = ["a", "b", "c", "d"]
{key: 0 for key in dict.fromkeys(keys).keys()}
```

```python
x = {
    "a" : 10, 
    "b" : 20, 
    "c" : 30, 
    "d" : 40
}

{value: 0 for value in x.values() }
```

```python
# # 딕셔너리 표현식에서 if조건문 사용하기 
x = {
    "a" : 10, 
    "b" : 20, 
    "c" : 30, 
    "d" : 40
}


for key, value in x.items():
    if value == 20: 
        del x[key]

print(x)



```

```python
a = [1, 2, 3, 4]
for i in a: 
    if i % 2 == 0:  
        a.remove(i)


print(a)
```

```python
x = {
    "a" : 10, 
    "b" : 20, 
    "c" : 30, 
    "d" : 40
}

{key: value for key, value in x.items() if value != 20 }
```

```python
# 딕셔너리의 할당과 복사 
x = {"a": 0, "b": 0, "c": 0, "d": 0}
y = x
```

```python
x is y
print(id(x))
print(id(y))
```

```python
y["a"] = 99
print(x)
print(y)
```

```python
x = {"a": 0, "b": 0, "c": 0, "d": 0}
y = x.copy()

x is y

y['a'] = 99

print(x)
print(y)
```

```python
x = {"a" : {"python": "2.7"}, "b": {"python" : "3.6"}}
y = x.copy()

y["a"]["python"] = "2.7.15"

print(x)
print(y)
```

```python
import copy

x = {"a" : {"python": "2.7"}, "b": {"python" : "3.6"}}
y = copy.deepcopy(x)

y["a"]["python"] = "2.7.15"

print(x)
print(y)
```
