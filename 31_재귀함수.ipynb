```python
def hello(count): 
    if count == 0: 
        return 50
    
    print("hello world", count)

    count -= 1
    return hello(count)
```

```python
print(hello(2))
```

```python
# 재귀호출로 팩토리얼 구하기

def factorial(n): 
    if n == 1: 
        return 1
    return n * factorial(n-1)
```

```python
6 * factorial(5)
```

```python
factorial(5)
```
